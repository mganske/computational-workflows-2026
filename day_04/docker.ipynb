### What is a container?

A container is an isolated process that packages everything needed to run a specific part of your application—such as a frontend, backend, or database—into a self-contained environment. Containers are:

**Self-contained:** They include all dependencies required to run, with no reliance on what's installed on the host machine. \
**Isolated:** Each container runs separately from others and from the host, minimizing conflicts and increasing security. \
**Independent:** Containers can be managed individually; deleting one doesn't affect others. \
**Portable:** Containers can run consistently across different environments, from your local machine to data centers or the cloud. \
\
This makes containers ideal for developing, testing, and deploying applications reliably and efficiently.

### Why do we use containers?

Containers are used because they provide a lightweight, consistent, and isolated environment to package and run applications. Key reasons for using containers include:

**Isolation and Security:** Containers run in loosely isolated environments, allowing multiple containers to run simultaneously on the same host without interfering with each other. \
**Portability:** Containers include everything needed to run an application, so they work the same way across different environments—on a developer’s laptop, in data centers, or in the cloud. \
**Fast, Consistent Delivery:** Containers streamline development, testing, and deployment by ensuring that applications behave the same way everywhere. This is especially useful for CI/CD workflows. \
**Efficient Resource Usage:** Containers are lightweight and use fewer resources compared to traditional virtual machines, allowing you to run more workloads on the same hardware. \
**Simplified Collaboration:** Developers can share containers, ensuring everyone works with the same environment and dependencies. \
\
These benefits make containers ideal for modern application development, testing, and deployment.


### What is a docker image?

 Docker image is a read-only template that contains the instructions for creating a Docker container. Images are often based on other images, with additional customizations such as installed software, application code, and configuration details. You can create your own images using a Dockerfile, which defines the steps to build the image. Each instruction in a Dockerfile creates a layer in the image, making images lightweight, small, and fast compared to other virtualization technologies.
