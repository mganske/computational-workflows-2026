#### 1. **Quality-control steps**
The nf-core/rnaseq pipeline uses tools such as FastQC and MultiQC to assess sequencing quality. Typical checks include:
- **Read quality:** identifies low-quality sequencing bases.
- **Adapter content:** detects leftover sequencing adapters.
- **GC content:** checks the proportion of guanine and cytosine bases.
- **Sequence duplication:** measures how often reads are repeated.
- **rRNA contamination:** assesses whether reads originate from ribosomal RNA.
- **Read and alignment statistics:** assess how many reads map successfully to the reference genome.

**Am I happy with the quality?**\
This cannot be determined without examining the actual FastQC and MultiQC reports. Ideally, reads have high base quality, low adapter contamination, and mapping statistics consistent with the experiment.

#### 2. **Additional information**
- **Ribosomal rRNA:** Ribosomal RNA is abundant and usually not the primary target in mRNA sequencing. High rRNA levels can indicate inefficient rRNA depletion or poly-A selection, reducing the reads available for studying gene expression.
- **Duplication:** High duplication can indicate PCR amplification bias or low library complexity. However, highly expressed genes can naturally produce many identical reads, so duplication is not always a sign of poor quality.
- **GC content:** GC content varies naturally between organisms and genomic regions. An unusual distribution or strong deviation from expectations can indicate contamination, amplification bias, or library-preparation problems.

#### 3. **What could lead to poorer results?**
- Degraded or low-quality RNA before library preparation.
- Inefficient rRNA depletion or poly-A selection.
- Contamination or incorrect sample handling.
- Excessive PCR amplification, causing high duplication.
- Adapter contamination or poor read trimming.
- Incorrect reference genome or annotation.
- Poor sequencing depth or unsuitable library preparation. \
These issues can reduce usable reads and make gene-expression estimates less reliable.
