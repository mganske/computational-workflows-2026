1. What is nf-core?

A community that builds curated, standardised and documented pipelines. They are all built with Nextflow and released under the MIT license. It also provides a template, shared modules and command-line tools.

2. How many pipelines are there currently in nf-core?

total: 206, released: 105

3. Are there any non-bioinformatic pipelines in nf-core?

Very few. The clearest is nf-core/rangeland, a pipeline for satellite imagery that produces trend files for land-cover change. Some others process images (mcmicro, molkart), not sequences. 

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

A series of processing steps in which each step uses the previous step’s output, written down so it can be rerun. It replaces slow, error-prone manual work with a repeatable, parallel and shareable analysis.

5. Why do you think nf-core adheres to strict guidelines?

So pipelines from many authors look and behave alike, modules can be shared, and quality, documentation and reproducibility are enforced automatically (linting, CI). The community can then maintain them, not a single person.

6. What are the main features of nf-core pipelines?

- A standard template, so every pipeline is structured alike.
- Software in containers (or Conda) with versioned tags, for portability and reproducibility.
- A parameter schema that validates inputs, including the samplesheet, before anything runs, and generates the `--help` text.
- Shared, versioned modules and subworkflows.
- Test profiles with small datasets, plus automated tests (CI, nf-test).
- Standard reporting: a MultiQC report, execution reports and software versions.
- Documentation for usage and outputs, and a community review process.
- Tooling (`nf-core` CLI) to list, launch, download and lint pipelines.
