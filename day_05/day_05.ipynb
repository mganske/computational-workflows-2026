### 1. Describe the concept of Workflows, Processes and Channels we deal with in Nextflow

**Workflows:** \
A workflow is the top-level structure that wires processes together via channels into an actual pipeline. It defines the overall logic, which processes run on which data, in what order, with what connections, essentially the blueprint for how data moves from raw input to final output (like the differential expression report). A workflow can itself call other, smaller workflows (subworkflows), which is how nf-core builds reusable, modular pipeline components.

**Processes:** \
A process is the basic unit of work in Nextflow, it defines a single computational step, such as running FastQC on a file, aligning reads with an aligner, or calling `fasterq-dump` to download data. Each process specifies:

- **input:** what data it needs (files, values)
- **output:** what it produces
- **script:** the actual shell/command block to execute

Processes are isolated from each other, they don't share state directly. Each time a process runs on a given input, Nextflow executes it in its own task directory, optionally inside a container, which is what gives Nextflow pipelines their portability and reproducibility.

**Channels:** \
Channels are how data flows between processes, they're essentially asynchronous queues that carry data (file paths, values, tuples of metadata plus files) from one process's output to another process's input. Rather than you manually scripting "run step A on all files, then once that's done run step B," you define a process's inputs as coming from a channel, and Nextflow handles scheduling, parallelization, and execution order automatically based on when data becomes available in that channel.

Channels are what enable the parallel, non-blocking execution that makes Nextflow efficient, for example, if ten samples are being processed, each one flows through the pipeline somewhat independently via channels, so a slow sample doesn't necessarily block others from proceeding through earlier steps.
